# 5. Hopfield: odtwarzanie wzorca

**Cel:** zapisać wzorzec w wagach, zakłócić jeden bit i śledzić energię podczas aktualizacji asynchronicznych. Czas: 15 min. Wymaga `numpy`.

**Uruchamianie:** wykonuj komórki kolejno. Dane są generowane lokalnie; nie jest potrzebny internet.

### Pamięć asocjacyjna
Stany są w `{−1,+1}`, macierz wag to iloczyn zewnętrzny jednego wzorca ze skasowaną przekątną. Przy aktualizacji jednego neuronu energia nie rośnie. Wzorzec i jego negacja są stanami stabilnymi.

```python
import numpy as np
pattern=np.array([1,1,-1,-1])
W=np.outer(pattern,pattern)/len(pattern)
np.fill_diagonal(W,0)
def energy(state): return float(-0.5*state@W@state)
def step(state,i):
    state=state.copy()
    field=float(W[i]@state)
    if field!=0: state[i]=1 if field>0 else -1
    return state
noisy=pattern.copy();noisy[2]*=-1
state=noisy.copy();trace=[energy(state)]
for _ in range(4):
    for i in range(4):
        state=step(state,i);trace.append(energy(state))
print('Zaszumiony:',noisy,'odtworzony:',state)
print('Energia:',np.round(trace,3))
assert np.array_equal(state,pattern)
assert all(a+1e-10>=b for a,b in zip(trace,trace[1:]))
```

**Analiza:** Zmień dwa bity. Sprawdź, czy wynik nadal jest oryginalnym wzorcem. Kiedy pamięć błędnie odtwarza inny atraktor? Atraktory opisuje teoria układów dynamicznych.
