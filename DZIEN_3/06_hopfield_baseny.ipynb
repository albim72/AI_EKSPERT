### Dynamika i stabilność
Aktualizujemy neurony po kolei. Remis zachowuje aktualną wartość; dzięki temu przebieg jest deterministyczny. Porównujemy rozmiary basenów różnych stanów końcowych.

```python
import numpy as np
from itertools import product
from collections import Counter
p=np.array([1,1,-1,-1]); W=np.outer(p,p)/4
np.fill_diagonal(W,0)
def E(x): return float(-0.5*x@W@x)
def relax(x):
    x=np.array(x,dtype=int); energies=[E(x)]
    for sweep in range(12):
        old=x.copy()
        for i in range(4):
            field=float(W[i]@x)
            if field: x[i]=1 if field>0 else -1
            energies.append(E(x))
        if np.array_equal(old,x):break
    assert all(a+1e-10>=b for a,b in zip(energies,energies[1:]))
    return tuple(x),len(energies)-1
basins=Counter()
for start in product((-1,1),repeat=4):
    target,steps=relax(start)
    basins[target]+=1
for target,size in basins.most_common():
    print('atraktor',target,'liczba startów',size,'energia',E(np.array(target)))
assert sum(basins.values())==16
assert tuple(p) in basins and tuple(-p) in basins
```
