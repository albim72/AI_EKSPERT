# 4. Inspiracja kwantowa: próbkowanie i aktualizacja

**Cel:** zobaczyć aktualizację rozkładu bitów za pomocą kątów. Jest to heurystyka na zwykłym CPU, bez kubitów ani przewagi kwantowej. Czas: 15 min. Wymaga `numpy`.

**Uruchamianie:** wykonuj komórki kolejno. Dane są generowane lokalnie; nie jest potrzebny internet.

### Kąt i prawdopodobieństwo
Dla bitu `P(1)=sin²(θ)` oraz `P(0)=cos²(θ)`. Wzór na amplitudy jest inspiracją; poniższy generator używa klasycznego RNG. Maksymalizujemy wartość przy budżecie masy 7.

```python
import numpy as np
rng=np.random.default_rng(29)
weights=np.array([2,3,4,5,1]);values=np.array([3,5,7,8,1]);capacity=7
def objective(bits):
    weight=int(bits@weights)
    return int(bits@values) if weight<=capacity else -10-weight
angles=np.full(5,np.pi/4)
assert np.allclose(np.sin(angles)**2,0.5)
seen=[]
for iteration in range(18):
    p=np.sin(angles)**2
    candidate=(rng.random(5)<p).astype(int)
    seen.append((objective(candidate),candidate.copy()))
    elite=max(seen,key=lambda z:z[0])[1]
    angles=np.clip(angles+0.08*(2*elite-1),0.05,np.pi/2-0.05)
    if iteration%6==5: print(iteration+1,'p(1)=',np.round(np.sin(angles)**2,2),'best=',max(seen,key=lambda z:z[0])[0])
print('Najlepsze znalezione:',max(seen,key=lambda z:z[0]))
assert np.all((np.sin(angles)**2>=0)&(np.sin(angles)**2<=1))
```

### Uczciwe porównanie budżetu
Losowe próbkowanie otrzymuje tyle samo prób. Jeden seed jest ilustracją, a nie statystycznym dowodem przewagi.

```python
random_rng=np.random.default_rng(30)
random_scores=[objective((random_rng.random(5)<0.5).astype(int)) for _ in range(len(seen))]
print('Heurystyka:',max(s for s,_ in seen),'losowo:',max(random_scores),'prób:',len(seen))
assert len(random_scores)==len(seen)==18
```

**Analiza:** Zwiększ krok aktualizacji 0,08 do 0,3. Czy prawdopodobieństwa zbyt szybko zbliżają się do 0 lub 1? Słowo „kwantowy” oznacza tu inspirację wzorem, nie wykonanie obliczeń na komputerze kwantowym.
