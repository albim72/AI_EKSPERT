### Reguły sterowania
Sygnał regulatora to 0–100%. Reguła 1: chłodno → moc 20; reguła 2: umiarkowanie → moc 50; reguła 3: gorąco → moc 90. Funkcje nakładają się, dzięki czemu odpowiedź zmienia się płynnie.

```python
import numpy as np
def triangle(x,a,b,c):
    if x<=a or x>=c:return 0.0
    if x==b:return 1.0
    return (x-a)/(b-a) if x<b else (c-x)/(c-b)
def controller(temp):
    cold=float(np.clip((25-temp)/15,0,1))
    medium=triangle(temp,15,25,35)
    hot=float(np.clip((temp-25)/15,0,1))
    strength=np.array([cold,medium,hot])
    if strength.sum()==0:raise ValueError('Brak aktywnych reguł')
    return float(strength@np.array([20,50,90])/strength.sum()),strength
for temp in (10,20,25,30,40):
    output,strength=controller(temp)
    print('t=',temp,'przynależności=',np.round(strength,2),'moc=',round(output,1))
assert controller(10)[0]==20
assert controller(25)[0]==50
assert controller(40)[0]==90
```

### Ciągłość w pobliżu progu
Porównaj wartości tuż po dwóch stronach 25°C; przy ostrym przełączniku byłby skok.

```python
left=controller(24.9)[0];right=controller(25.1)[0]
print('24.9°C:',round(left,3),'25.1°C:',round(right,3),'różnica:',round(right-left,3))
assert abs(right-left)<2
```

**Analiza:** Przesuń środkową regułę do 27°C i obejrzyj reakcję. Dlaczego trzeba zdefiniować postępowanie, gdy żadna reguła się nie uaktywnia? To demonstracja, a nie gotowy regulator urządzenia.
