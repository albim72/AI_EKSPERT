# 7. Logika rozmyta: alarm dla pomiarów

**Cel:** zastąpić ostry próg stopniowalną oceną wysokiej temperatury i drgań. Czas: 12 min. Wymaga `numpy`.

**Uruchamianie:** wykonuj komórki kolejno. Dane są generowane lokalnie; nie jest potrzebny internet.

### Funkcje przynależności
Temperatura jest „wysoka” od 60 do 90°C, drgania od 2 do 8 mm/s. Wartości od 0 do 1 oznaczają stopień przynależności, **nie** prawdopodobieństwo. Reguła: alarm = wysoka temperatura **LUB** silne drgania (`max`).

```python
import numpy as np
def ramp(x,low,high):return float(np.clip((x-low)/(high-low),0,1))
def alarm(temp,vibration):
    high_t=ramp(temp,60,90)
    strong_v=ramp(vibration,2,8)
    return max(high_t,strong_v)
measurements=[(55,1),(75,1),(55,5),(80,6),(95,9)]
for t,v in measurements: print(f'{t:3} C, {v:2} mm/s -> alarm {alarm(t,v):.2f}')
assert alarm(55,1)==0
assert alarm(95,9)==1
assert np.isclose(alarm(75,1),0.5)
```

### Koniunkcja
Reguła „gorąco **I** drga” daje `min` i jest bardziej restrykcyjna niż `max`.

```python
for t,v in measurements:
    both=min(ramp(t,60,90),ramp(v,2,8))
    print((t,v),'OR:',round(alarm(t,v),2),'AND:',round(both,2))
```

**Analiza:** Zmień przedział temperatury na 70–85°C. Dla jakich pomiarów zmienia się decyzja o alarmie? Przy rzeczywistym zastosowaniu zakresy i reguły wymagają uzgodnienia z ekspertami.
